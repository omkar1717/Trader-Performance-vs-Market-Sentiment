### Part - B

1. Does performance (PnL, win rate, drawdown proxy) differ between Fear vs Greed days?

Yes, trading performance differs significantly between Fear and Greed days. 
Analysis indicates that while "Greed" periods are marked by higher, steady, but risky gains,
"Fear" periods often provide better opportunities for contrarian traders to achieve higher average profits and win rates,
albeit with higher volatility and risk of large, sudden drawdowns. 

2. Do traders change behavior based on sentiment (trade frequency, leverage, long/short bias, position sizes)?

Yes, traders frequently change their behavior based on market sentiment, often shifting between rational analysis and emotional responses that impact trade frequency, leverage, long/short bias, and position sizes. Positive sentiment (optimism) tends to increase buying, while negative sentiment (fear) triggers selling or increased use of defensive, low-risk, or hedging strategies.

3. Identify 2–3 segments:

a. high leverage vs low leverage traders--

    high leverage-
    - Use higher leverage to control large positions with minimal capital.
    - They are usually short-term, intraday, or scalp traders
    
    low leverage traders-
    - Use minimal or no leverage 
    - They are long-term investors or those holding positions over longer periods

b. frequent vs infrequent traders--

    frequent-
    - Actively trade, often entering and exiting multiple positions in minutes or seconds.
    - They rely on high-speed execution, technical analysis, and small price fluctuations
    
    infrequent traders-
    - Trade less often, holding positions for days, weeks, or months.
    - They prioritize in-depth, fundamental, or long-term technical analysis.
    
c. consistent winners vs inconsistent traders--

    consistent winners-
    - Prioritize risk management above profit, often risking only 1-2% of their total capital on any single trade
    - They treat trading as a business rather than a hobby or gambling.
    
    inconsistent traders-
    - Often display behaviors such as "revenge trading" after losses, ignoring stop losses, or over-leveraging
    - They often trade on emotion or rumors rather than a tested, consistent strategy.

```python
daily_winrate = df2.groupby("date")["is_win"].mean()
```

```python
plt.figure()
plt.plot(daily_winrate.index, daily_winrate.values)
plt.xlabel("Date")
plt.ylabel("Daily win rate")
plt.title("Daily win rate")
plt.show()
```

```python
daily_pnl = df2.groupby("date")["Closed PnL"].sum()
```

```python
plt.figure()
plt.plot(daily_pnl.index, daily_pnl.values)
plt.xlabel("Date")
plt.ylabel("Daily total PnL")
plt.title("Daily total PnL over time")
plt.show()
```

```python
long_per_day = (
    df2[df2["Direction"].str.lower() == "long"]
    .groupby("date")
    .size()
)

short_per_day = (
    df2[df2["Direction"].str.lower() == "short"]
    .groupby("date")
    .size()
)

combined = pd.concat([long_per_day, short_per_day], axis=1)
combined.columns = ["long_trades", "short_trades"]
combined = combined.fillna(0)
```
